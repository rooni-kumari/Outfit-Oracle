Step 1: Upload Person Image

```python
print("👤 Please upload the PERSON image:")
person_upload = files.upload()

if len(person_upload) != 1:
    print("❌ Upload ONE person image.")
    raise SystemExit

person_img_name = list(person_upload.keys())[0]
person_img = Image.open(BytesIO(person_upload[person_img_name]))
print(f"✅ Loaded person image: {person_img_name}")
```

Step 2: Upload Wardrobe Images

```python
print("\n🧥 Please upload WARDROBE images (e.g., shirt, pants, shoes):")
wardrobe_upload = files.upload()

if len(wardrobe_upload) == 0:
    print("❌ Upload at least one wardrobe image.")
    raise SystemExit

wardrobe_imgs = []
wardrobe_img_names = []

for name in wardrobe_upload:
    try:
        img = Image.open(BytesIO(wardrobe_upload[name]))
        wardrobe_imgs.append(img)
        wardrobe_img_names.append(name)
        print(f"👕 Loaded wardrobe image: {name}")
    except:
        print(f"⚠️ Skipped invalid image: {name}")

```

Display images

```python
print("\n🖼️ Displaying images...")
person_img.show()
for img in wardrobe_imgs:
    img.show()
```

Step 3: User Input

```python
occasion = input("\n🎯 Enter the occasion (e.g., party, interview, wedding): ")
gemini_api_key = input("🔑 Enter your Google Gemini API key (starts with AIza...): ")
```

Step 4: Setup Gemini

```python
genai.configure(api_key=gemini_api_key)
model = genai.GenerativeModel("gemini-1.5-flash")  # ✅ Updated model name
```

Step 5: Create Prompt

```python
prompt = f"""
You're a professional fashion stylist.

A person is preparing for an occasion: "{occasion}".

Their wardrobe includes: {', '.join(wardrobe_img_names)}.

Choose the best outfit from the wardrobe for the occasion and explain why it is suitable.
"""

print("\n⏳ Generating outfit recommendation from Gemini...")

```

Step 6: Get Response (New API Syntax)


```python

inputs = [prompt, person_img] + wardrobe_imgs
response = model.generate_content(inputs)


print("\n👗 Outfit Recommendation:")
print(response.text)

```
